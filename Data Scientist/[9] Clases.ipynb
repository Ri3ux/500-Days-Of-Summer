# Clases y Objetos en Python

## ¿Qué es una clase?
Una **clase** es como un molde o plantilla para crear **instancias**, que son objetos individuales con las características definidas por la clase.

Por ejemplo, el string `'¡Hola, mundo!'` es una **instancia** de la clase str (cadena de texto).

Todas las instancias de una clase tienen los mismos **atributos y métodos**. En Python, existen muchas clases incorporadas, como `str` (strings), `dict` (diccionarios) y `list` (listas). Cada una tiene métodos específicos que permiten manipular sus valores.

Por ejemplo, el método `append()` se puede usar en cualquier lista porque está definido en la clase `list`:

```python
my_list = ['coat', 'goat']
my_list.append('boat')
print(my_list)
```

## Crear una clase
Supongamos que estamos desarrollando un videojuego y queremos crear algunos personajes con diferentes características, así como definir las acciones que el jugador pueda realizar en nuestro juego. Hay diferentes tipos de personajes: magos, guerreros, etc., los cuales tienen diferentes propiedades y pueden hacer distintas cosas.

¿Cómo convertiríamos el diseño de personajes en código Python? Una posible solución sería utilizar diccionarios para crear personajes y funciones para interactuar con el jugador, por ejemplo, así:

```python
mage = {'health': 50, 'damage': 10, 'knowledge': 95}
knight = {'health': 100, 'damage': 25, 'knowledge': 20}

arthur = knight.copy() # hacer una copia del diccionario 'knight' original
arthur['name'] = 'Arthur' # reemplazar el nombre dentro de la copia

richard = knight.copy() # hacer otra copia del diccionario 'knight' original
richard['name'] = 'Richard' # reemplazar el nombre dentro de la copia

def heal(character): # crear la función que cambia la salud
    character['health'] += 20

heal(richard) # llamar a la función para cambiar la salud de Richard

print(richard)
```

Pero parece medio engorroso y tedioso. 

Gracias a las clases podemos crear personajes de videojuegos con mayor facilidad. Veamos el mismo caso, pero usando clases:

```python
# Creamos la clase Knight
class Knight:
    def __init__(self, name):
        # Establecemos los parámetros
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

arthur = Knight('Arthur')
richard = Knight('Richard')

print(arthur)
```

Hemos creado una clase llamada `Knight()` usando la sentencia `class`. Luego definimos a cada uno de los caballeros, Arthur y Richard. 

El código con **clases** es mejor porque es más **organizado**, **reutilizable** y **fácil de mantener**.

1. **Menos repetición:** Con clases, creamos objetos sin copiar manualmente datos como en los diccionarios.
2. **Métodos propios:** Podemos agregar funciones dentro de la clase, como `heal()`, sin necesidad de escribir funciones separadas.
3. **Más escalable:** Si queremos agregar más personajes (`Mage`, `Archer`), solo creamos nuevas clases en lugar de manejar múltiples diccionarios.

Usar **clases** hace que el código sea más limpio, fácil de entender y modificar. 🚀

Aunque Python no lo requiere, lo habitual es que los nombres de las clases empiecen siempre con mayúscula y utilicen camel case en lugar de guion bajo si el nombre de la clase está formado por más de una palabra.

Todas las clases de Python tienen un método especial `__init__()` al que se llama cuando se crea una instancia de la clase. Por ejemplo, cuando usamos `Knight('Arthur')` para crear la variable `arthur`. El parámetro especial `self~~=~~` se refiere a la instancia que estamos creando, por lo que no tenemos que pasarle ningún argumento. Sin embargo, tenemos que especificar el argumento `name~~=~~` cada vez que creamos una instancia de `Knight()` porque no hemos especificado un valor por defecto.

## Atributos
Cada instancia de `Knight()` tiene atributos a los que podemos acceder utilizando la notación de punto:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

arthur = Knight('Arthur')
print(f'La salud de {arthur.name} es {arthur.health}')
print(f'El daño de {arthur.name} es {arthur.damage}') 
print(f'La inteligencia de {arthur.name} es {arthur.knowledge}') 
```

Un atributo es un valor particular de esa instancia de clase. Es decir, la instancia `arthur` tendrá un valor del atributo `damage` específico (25). 

También podemos cambiar el valor de cualquier atributo simplemente asignándole un nuevo valor:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

arthur = Knight('Arthur')
print(f'El daño inicial de {arthur.name} es {arthur.damage}')

arthur.damage = 50
print(f'El nuevo daño de {arthur.name} es {arthur.damage}')
```

Por último, podemos ver todos los atributos de una instancia en forma de diccionario a través del atributo especial `__dict__`:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

arthur = Knight('Arthur')

print(f'{arthur.name} tiene estos atributos {arthur.__dict__}')
```

## Métodos
Nuestros caballeros tienen algunos atributos útiles, pero ¿qué pasa si queremos que actúen? Podemos definir, dentro de una clase, funciones personalizadas que estarán disponibles para que las instancias de la clase las utilicen. Estas funciones específicas de una clase se llaman **métodos**:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

    def heal(self):
        self.health += 20

    def learn(self):
        self.knowledge += 5
```

Podemos definir métodos dentro de una clase exactamente de la misma manera que lo haríamos con cualquier otra función. Los métodos `heal()` (sanar) y `learn()` (aprender) tienen un único parámetro `self~~=~~`, lo que significa que la llamada a estos métodos afecta a la instancia que los llama; no tenemos que pasarles ningún argumento. Llamamos a los métodos utilizando la famosa notación de puntos:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

    def heal(self):
        self.health += 20

    def learn(self):
        self.knowledge += 5

arthur = Knight('Arthur')

arthur.heal()
arthur.learn()

print(f'{arthur.name} se ha curado. Su salud es de {arthur.health}')
print(f'{arthur.name} ha aumnetado su inteligencia. Su inteligencia es de {arthur.knowledge}')
```

El atributo `health` se incrementó en 20 y el atributo `knowledge` en 5, tal y como especificamos en `heal()` y `learn()`, respectivamente. Si queremos que nuestros métodos sean más flexibles, podemos incluir otros parámetros:

```python
class Knight:
    def __init__(self, name):
        self.health = 100
        self.damage = 25
        self.knowledge = 20
        self.name = name

    def heal(self, amount):
        self.health += amount

    def learn(self, amount):
        self.knowledge += amount

arthur = Knight('Arthur')

# Aumentamos 20 puntos de salud
arthur.heal(50)

# Aumentamos 3 puntos de inteligencia
arthur.learn(4)

print(f'{arthur.name} se ha curado. Su salud es de {arthur.health}')
print(f'{arthur.name} ha aumnetado su inteligencia. Su inteligencia es de {arthur.knowledge}')
```

## Métodos estáticos
Hasta ahora, todos los métodos que vimos afectan a la instancia de la clase que llama al método; incluso el método especial `__init__()` incluye el parámetro `self`. Los **métodos estáticos** no están vinculados a una instancia específica de una clase ni requieren el parámetro `self`. Esto significa que pueden ser llamados sin crear un objeto de esa clase en particular. Además, los métodos estáticos no tienen la capacidad de modificar el estado de un objeto, ya que no tienen acceso directo a las propiedades de la instancia.

Para crear un método estático, tenemos que utilizar el **decorador** `@staticmethod`.

A continuación, tenemos un ejemplo de la clase `Stock()` que rastrea información sobre los tickers de las acciones. El método `show_current_price` tiene el objetivo de obtener el precio actual de las acciones de Internet e imprimirlo.

```python
class Stock:
    # Precios simulados (atributo de clase)
    prices = {
        "AAPL": 227.50,
        "MSFT": 430.10,
        "TSLA": 251.30,
    }
    def __init__(self, ticker, amount):
        self.ticker = ticker
        self.amount = amount

    @staticmethod
    def show_current_price(ticker):
        current = Stock.prices.get(ticker, "Ticker no encontrado")
        print(f'{ticker}: {current}')

# Uso: no necesitas crear un objeto
Stock.show_current_price("AAPL")

# También funciona desde una instancia
my_asset = Stock("MSFT", 10)
my_asset.show_current_price("MSFT")
```

Cuando llamas a `Stock.show_current_price('AAPL')`, simplemente invocas el método estático `show_current_price` de la clase `Stock`. No se crea una instancia de la clase `Stock`, así que no se crea un objeto. El método estático puede ser llamado directamente desde la clase sin crear una instancia de la clase. En este caso, `'AAPL'` se pasa como un argumento al método estático, pero no se crea un objeto `Stock` en el proceso.

## Métodos de clase
Los **métodos de clase** se pueden utilizar para implementar formas alternativas de crear instancias. Un método de clase toma un objeto de clase como el primer argumento. Al crear un método de clase, es necesario utilizar el decorador `@classmethod`.

Los **métodos de clase** permiten crear objetos de formas alternativas. Para usarlos, se define `@classmethod`, y el primer argumento es la clase `(cls)`.

```python
class Stock:
    def __init__(self, ticker, amount, price):
        self.ticker = ticker
        self.amount = amount
        self.price = price
        self.total = self.amount * self.price

    @classmethod
    def from_string(cls, string):
        ticker, amount, price = string.split()
        return cls(ticker, int(amount), float(price))

# Crear instancia desde un string
abc = Stock.from_string('ABC 10 1.5')

# Forma normal de crear el objeto
xyz = Stock('XYZ', 5, 2.0)
print(f'{xyz.ticker}: {xyz.amount} acciones a ${xyz.price} = ${xyz.total}')

# Crear el objeto desde un string
abc = Stock.from_string("ABC 10 1.5")
print(f'{abc.ticker}: {abc.amount} acciones a ${abc.price} = ${abc.total}')
```

Esto nos permite crear objetos fácilmente desde datos en texto sin escribir código repetitivo.

## Herencia
Volvamos al código de nuestro videojuego. Escribimos una clase `Knight()` fantástica, pero ahora queremos añadir más clases, por ejemplo, magos, campesinos, etc.

Al fin y al cabo, las clases sirven para escribir código reutilizable. Pues bien, podemos utilizar la **herencia** para crear todas las clases de personajes que necesitemos a partir de una única clase padre, en lugar de crear cada clase desde cero.

Vamos a construir una clase padre `Character()` y usarla para crear sus clases hijas, `Knight()` y `Peasant()`:

```python
# Construimos la clase padre
class Character:
    def __init__(self, name):
        self.name = name
        self.health = 100

    def heal(self, value:int=20):
        self.health += value

    def learn(self, value:int=20):
        self.knowledge += value

# Construimos la clase hijo de Knight añadiendo atributos específicos
class Knight:
    def __init__(self, name):
        Character.__init__(self, name) # Se hereda el constructor de la clase padre
        self.damage = 25 # Añadimos el nuevo atributo
        self.knowledge = 20 # Añadimos el nuevo atributo

# Construimos la clase hijo de Peasant añadiendo atributos específicos
class Peasant:
    def __init__(self, name):
        super().__init__(self, name) # Otra forma de heredar el constructor de la clase padre
        self.damage = 10 # Añadimos el nuevo atributo
        self.knowledge = 36 # Añadimos el nuevo atributo

arthur = Knight('Arthur')
richard = Peasant('Richard')

print(f'Los atributos de {arthur.name} son {arthur.__dict__}')
print(f'Los atributos de {richard.name} son {richard.__dict__}')
```

Hemos creado dos clases hijas: `Knight` y `Peasant`, ambas heredan todos los métodos y propiedades de la clase padre `Character`. Observa que la herencia se implementa de diferente forma entre las clases `Knight` y `Peasant`: para la clase `Character` utilizamos la construcción `Character.__init__(self, name)`. En cambio `super()` automáticamente determina la clase padre correcta y llama a su constructor, lo que significa que no tienes que especificar explícitamente el nombre de la clase padre como hicimos cuando construimos la clase `Knight`.

La clase `Knight()` funciona utilizando la herencia de `Character()`. Podemos reutilizar fácilmente `Character()` para construir otras clases con características comunes a `Knight()` para crear en el juego cualquier otro personaje a nuestro gusto. Por ejemplo, todas las clases de personajes tienen un nombre y un nivel de salud, que son atributos de `Character()`.

4. Supongamos que necesitamos una forma rápida de añadir una cuenta utilizando una cadena que almacene las ID de la cuenta y del titular, por ejemplo, tenemos una cadena `"001/2406"`. Para lograrlo, debes crear el método de clase `quick()`. Este método debe esperar una cadena con las ID de cuenta y titular, separarlas y almacenar cada valor correspondiente en las variables `acc_id` y `holder_id`. Luego, el método debe devolver la clase, donde bank siempre es `default_bank`, seguido de `acc_d`, `holder_id` y, finalmente, un `balance` de `0.0`, en ese orden.

```python
from datetime import datetime

class Account:
    def __init__(self, bank, acc_id, holder_id, balance:float=0.0):
        self.bank = bank
        self.acc_id = acc_id
        self.holder_id = holder_id
        self.balance = balance
        self.start_date = datetime.now()

    def deposit(self, amount:float):
        self.balance += amount
        
    def withdraw(self, amount:float):
        self.balance -= amount

    @staticmethod
    def bankphone(bank):
        '''
        imprimir el número del banco
        '''
        print(f'El número de teléfono de {bank} es 1-000-1234567')

    @classmethod
    def quick(cls, string):
        '''
        crea una cuenta a partir de una cadena
        usando solo las identificaciones de cuenta y titular
        separadas por una barra
        '''
        acc_id, holder_id = string.split('/') # Si lo dejamos vacío, tomará el espacio como separador
        return cls('default_bank', acc_id, holder_id, 0.0)
```

```python
from datetime import datetime

class Account:
    def __init__(self, bank, acc_id, holder_id, balance:float=0.0):
        self.bank = bank
        self.acc_id = acc_id
        self.holder_id = holder_id
        self.balance = balance
        self.start_date = datetime.now()

    def deposit(self, amount:float):
        self.balance += amount
        
    def withdraw(self, amount:float):
        self.balance -= amount

    @staticmethod
    def bankphone(bank):
        '''
        imprimir el número del banco
        '''
        print(f'El número de teléfono de {bank} es 1-000-1234567')

    @classmethod
    def quick(cls, string):
        '''
        crea una cuenta a partir de una cadena
        usando solo las identificaciones de cuenta y titular
        separadas por una barra
        '''
        acc_id, holder_id = string.split('/') # Si lo dejamos vacío, tomará el espacio como separador
        return cls('default_bank', acc_id, holder_id, 0.0)

first = Account('old_trusty', '001', '10043', 500)
first.deposit(250)
first.withdraw(400)
print(f'El saldo de la primera cuenta es: ${first.balance}')

second = Account.quick('002/10123')
print(f'El año de creación de la segunda cuenta es: {second.start_date.year}')
```
