# Calidad del código
Un código de buena calidad es fácil de leer, comprender y mantener, funciona bien cerca de las condiciones límite y utiliza los recursos informáticos de manera eficiente. Todo código de calidad debe tener los siguientes factores:

* Estilo de código
* Confiabilidad
* Rendimiento
* Seguridad

La calidad del código es importante por muchas razones, especialmente a medida que aumenta la cantidad de personas que utilizan productos impulsados por ese código, así como la cantidad de desarrolladores y desarrolladoras que trabajan en el mismo:

* **Comunicación:** otras personas pasan menos tiempo tratando de entender lo que hace el código.
* **Mantenimiento:** es más fácil encontrar y corregir errores, así como ampliar la funcionalidad del código.
* **Previsibilidad:** funciona como se esperaba y minimiza las sorpresas desagradables cuando las cosas se detienen de forma repentina.

## Estilo del código
El **estilo del código** se refiere a cómo se ve el código, no a cómo funciona. Así como existen reglas para una comunicación clara y concisa en los lenguajes naturales, también tenemos reglas de estilo de código para los lenguajes de programación. Aquí tienes algunos ejemplos de reglas de estilo en Python:

* El número de espacios para la indentación;
* El estilo y denominación de las variables, por ejemplo, `book_title` vs. `BookTitle`
* Poner espacios alrededor de un operador de asignación (`a = 1`) y omitirlos para los argumentos al llamar a una función (`my_function(a=1)`)

Un conjunto de este tipo de reglas se conoce comúnmente como una `guía de estilo`.

Diferentes personas u organizaciones pueden adherirse a diferentes reglas, formando efectivamente diferentes guías de estilo. A veces, más de una guía de estilo se vuelve popular, por ejemplo, las grandes empresas de tecnología con mucho desarrollo interno, como Google, tienden a establecer sus propias [guías de estilo internas](https://developers.google.com/style). Para Python, la mayoría de las personas se adhieren en gran medida a la guía de estilo [PEP 8](https://peps.python.org/pep-0008/), por lo que debes consultarla si alguna vez tienes dudas sobre cómo diseñar tu propio código.

## Hacer cumplir el estilo de código
Incluso con una guía de estilo completa, es común que las personas se desvíen de ella aquí y allá. Existen herramientas que pueden ayudar a validar el código para comprobar si se ajusta a un estilo de código particular y alertarnos si hay algún error. Estas herramientas se llaman **linters**. Los linters pueden comprobar diferentes aspectos del estilo de código, algunos son más restrictivos, mientras que otros son más adecuados para proyectos de gran escala. Cualquier linter debería ser suficiente para proyectos pequeños. Los linters populares se integran fácilmente con los IDE.

[Pylint](https://pylint.pycqa.org/en/latest/) es un linter para Python, bien establecido y de buena reputación. Puedes instalarlo con pip como una herramienta de línea de comandos

````python
# Instalamos el linter
%pip install pylint
````

Y ejecutarlo para comprobar el estilo de código de un script con:

````python
pylint # Ingresa la ruta de tu script
````

## Código de prueba
La prueba del producto es la práctica de validar que un producto funcione como se esperaba. Esta práctica se ha adoptado en muchas industrias para minimizar los riesgos de errores inesperados. Aprenderemos técnicas comunes para realizar pruebas en Python: **pruebas unitarias** con `pytest` y la herramienta de línea de comandos.

Las **pruebas unitarias** se utilizan para probar partes individuales de nuestro código. La librería `pytest` es ideal para ejecutar pruebas unitarias en Python, ya que ofrece funcionalidades como:

* Ejecutar pruebas explícitamente;
* Filtrar qué casos de prueba ejecutar;
* Volver a ejecutar solo las pruebas fallidas;
* Informar los resultados de las pruebas.

Puedes instalar pytest con pip como una herramienta de línea de comandos:

````python
# Instalamos el módulo de pruebas unitarias
%pip install pytest
````

En un escenario básico, `pytest` busca en tu código funciones con nombres que comiencen con `test_`, y ejecuta esas funciones al encontrarlas. Si no se genera una `AssertionException` dentro de una función de prueba, se considera aprobada:

````python
# Crea la función sign()
def sign(x):
    """Devuelve el signo de número."""
    if x == None:
        return None
    if x < 0:
        return -1
    return 1

# Prueba la función sign()
def test_sign():
    assert sign(-10) == -1
    assert sign(10) == 1
    assert sign(0) == 1
    assert sign(None) == None
````

`AssertionException` es un tipo de excepción que se lanza con declaraciones de aserción, las cuales son declaraciones lógicas de prueba de cordura insertadas en el código. Estas funciones están destinadas a devolver siempre `True` a menos que haya un problema. La sintaxis es:

```python
assert expression, "mensaje de aserción"
```

La expression debe producir `True` si no hay ningún problema, de lo contrario, `False`. En este último caso, la excepción `AssertionError` se lanza con un método de aserción si se proporciona uno.

La función `test_sign()` anterior es una prueba unitaria que contiene varias afirmaciones, cada una de las cuales comprueba su propio conjunto de condiciones.

Si guardamos este script como `unite_test.py`, podemos ejecutar pruebas unitarias con `pytest`:

````python
import sys
!{sys.executable} -m pytest unite_test.py
````

Escribir pruebas unitarias integrales puede ser un desafío, ya que requiere algo de imaginación para descubrir todas las formas posibles en las que un usuario puede descifrar tu código, así como también cómo probar esas situaciones. Tener personas de control de calidad (QA) es una práctica estándar en muchas industrias; el trabajo de estas personas es cubrir las funcionalidades críticas de un programa con pruebas unitarias.

Al ejecutar una prueba unitaria (`-rA` significa que pedimos a `pytest` que muestre todos los resúmenes de las pruebas).

````python
import sys
!{sys.executable} -m pytest -rA unite_test.py
````
