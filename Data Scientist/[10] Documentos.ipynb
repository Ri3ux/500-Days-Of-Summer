## Archivos texto o markdown

### Lectura de archivos
El método `read()` lee todo el archivo como una sola cadena, mientras que el método `readlines()` crea una lista de cadenas, donde cada elemento de la lista es una línea del archivo:

````python
f = open('Ulysses-Alfred, Lord Tennyson.md', mode='r', encoding='utf-8') # Cambianmos la codificación para poder leer texto en español

print(f.read())
````

````python
with open('Ulysses-Alfred, Lord Tennyson.md', mode='r', encoding='utf-8') as f:
    for line in f.readlines():
        print(line.strip())
````

### Escritura de archivos
Tenemos dos opciones que podemos pasar como argumento `mode=` a `open()` para escribir:

* `'w'`: w de write (en inglés, escribir), abre un nuevo archivo vacío en el que se puede escribir.
* `'a'`: a de append (en inglés, anexar), abre un archivo nuevo o ya existente y le añade texto.

Ten cuidado al utilizar `mode='w'`: si abres un archivo con la misma ruta y nombre que un archivo existente, ¡sobre escribirás completamente el contenido de este antiguo archivo!

Después de abrir un archivo, debemos escribir datos en él. Lo podemos hacer con el método `write()`, que toma un único argumento de cadena y escribe esa cadena en cualquier lugar del archivo, comenzando por donde nos encontremos.

````python
from datetime import datetime

# Obtenemos la hora actual

now = datetime.now()

with open('Ulysses-Alfred, Lord Tennyson.md', mode='a', encoding='utf-8') as f:
    f.write('\n') # Comenzamos con una nueva línea
    f.write(str(now)) # Añadimos la hora actual
    f.write('\n\n') # Añadimos un espacio vacío
    f.write(input()) # Escribimos una entrada en el archivo desde el prompt
    f.write('\n\n') # Termimanos con una línea en blanco
````

Si tenemos una lista de cadenas y queremos escribir por orden cada cadena en el archivo, podemos utilizar el método `writelines()`:

````python
ulyses = [
            '> Aunque mucho se ha gastado mucho queda aún; y si bien\n'
            '> no tenemos ahora aquella fuerza que en los viejos tiempos\n'
            '> movía tierra y cielo, somos lo que somos:\n'
            '> corazones heroicos de parejo temple,\n'
            '> debilitados por el tiempo y el destino, más fuertes en voluntad\n'
            '> para esforzarse, buscar, encontrar y no rendirse.'
        ]

with open('Ulysses-Alfred, Lord Tennyson.md', mode='a', encoding='utf-8') as f:
    f.writelines(ulyses)
````

Si queremos escribir cada cadena de la lista en su propia línea, cada una de ellas debe terminar con el carácter de nueva línea `\n`.

## Archivos JSON
JSON es un formato de almacenamiento y transferencia de datos universal que los humanos pueden leer, utilizado a menudo para manejar datos en aplicaciones web. Los archivos JSON son muy similares a los diccionarios de Python; se trata de un conjunto de pares `clave-valor`, que suelen tener muchas capas anidadas.

Podemos utilizar la librería incorporada `json` para trabajar fácilmente con archivos `.json`.

La convención `dump/loads` se utiliza cuando es necesario guardar/leer archivos `.json`.

### Lectura de archivos

````python
import json
import requests # Importamos la librería para obtener datos de internet

# Descargamos un archivo JSON de internet
get_data = requests.get('https://dummyjson.com/products/1')

# Extraemos el contenido del archivo JSON descargado

text = get_data.text

# Leemos el contenido del archivo
print(json.loads(text))
````

El texto parece tener la forma de una cadena; aunque se parece a un archivo JSON, se descarga como una cadena usando `requests.get()`. Por lo tanto, se requiere la función `json.loads()` para convertirlo (o analizarlo) en un archivo JSON real.

### Escritura de archivos

````python
import json

# Escribiremos este diccionario en un archivo JSON
data = {
    'user_id': 12,
    'user_name': 'Ransbach',
    'full_name': 'Ricardo Estepa',
    'email': 'sir.ransbach@gmail.com',
    'status': 'active',
    'is_premium': True,
    'created_at': '2026-10-12',
    'age': 37,
    'location': {
        'city': 'Bogotá',
        'country': 'Colombia',
        'timezone': 'America/Bogota'
    },
    'interests': ['Data Science', 'Rock Music', 'Photography'],
    'social_links': {
        'github': 'https://github.com/ri3ux',
        'linkedin': None
    },
    'stats': {
        'followers': 340,
        'following': 128,
        'posts': 57
    }
}

# Creamos una archivo JSON en modo escritura y volcamos nuestro datos allí

with open('profile.json', mode='w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
````

Para leer los datos anidados de un archivo JSON, hacemos lo siquiente:

````python
with open('profile.json', mode='r', encoding='utf-8') as f:
    profile = json.load(f)

print(f'Ciudad: {profile['location']['city']}')
print(f'Intereses: {profile['interests'][0]}')
print(f'Seguidores: {profile['stats']['followers']}')
````

### Ejercicio: filtrar productos de una marca

La URL `https://dummyjson.com/products/category/smartphones` devuelve, en formato JSON, un catálogo de teléfonos inteligentes. La respuesta es un diccionario cuya clave `'products'` contiene una lista de productos, y cada producto es a su vez un diccionario con datos como `'title'`, `'price'` y `'brand'`.

Tu objetivo es quedarte solo con los productos de la marca **Samsung** y guardarlos en un archivo.

1. Descarga los datos de la URL con la librería `requests` y extrae el cuerpo de la respuesta con el atributo `.text`.
2. Convierte ese texto en un objeto de Python usando la función adecuada de la librería `json`. Guarda el resultado en la variable `json_data`.
3. Extrae la lista de productos de `json_data` con la clave `'products'` y guárdala en la variable `products`.
4. Crea una lista vacía llamada `items` y una variable `brand` con el valor `'Samsung'`.
5. Recorre la lista `products`. Para cada producto, obtén su marca con la clave `'brand'` y, si coincide con el valor de `brand`, agrega el producto completo a la lista `items`.
6. Crea un archivo llamado `'samsung_items.json'` y guarda en él la lista `items`.

````python
import json
import requests

# obtener datos de internet
response = requests.get('https://dummyjson.com/products/category/smartphones')
text = response.text

json_data = json.loads(text)
products = json_data['products']

items = []
brand = 'Samsung'

for entry in products:
    if entry['brand'] == brand:
        items.append(entry)

with open('samsung_items.json', mode='w') as f:
    json.dump(items, f, indent=4, ensure_ascii=False)
````
