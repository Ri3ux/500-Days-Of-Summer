# Herramientas de desarrollo de software

## Comandos de Bash

| Comando | Significado | Descripción | Ejemplo |
| --- | --- | --- | --- |
| `alias` | *alias* (sobrenombre) | Crea un atajo para un comando. Sin argumentos, muestra los alias que existen. Solo dura mientras la terminal esté abierta; para que sea permanente, agrégalo a `~/.bashrc` | `alias ll='ls -la'` |
| `cat` | *concatenate* (concatenar) | Muestra el contenido de un archivo. Con varios archivos, los muestra uno tras otro | `cat notas.txt` |
| `cat -n` | *number* (numerar) | Muestra el contenido con el número de cada línea | `cat -n notas.txt` |
| `cat >` | Concatenar y guardar | Une varios archivos en uno nuevo | `cat a.txt b.txt > todo.txt` |
| `cd` | *change directory* (cambiar de directorio) | Avanza hacia alguna ubicación dentro del sistema | `cd Documents` |
| `cd ..` | `..` = un nivel arriba | Regresa a la carpeta anterior (la carpeta "padre") | `cd ..` |
| `cd ~` | `~` = *home* | Va a tu carpeta de usuario. Escribir solo `cd` hace lo mismo | `cd ~` |
| `cd -` | `-` = carpeta anterior | Regresa a la última carpeta donde estuviste | `cd -` |
| `cd /` | `/` = raíz | Va a la raíz del sistema, el nivel más alto | `cd /` |
| `cp` | *copy* (copiar) | Copia archivos | `cp notas.txt copia.txt` |
| `cp -r` | *recursive* (recursivo) | Copia una carpeta con todo su contenido | `cp -r datos respaldo` |
| `cp -i` | *interactive* (interactivo) | Pregunta antes de sobrescribir un archivo que ya existe | `cp -i notas.txt copia.txt` |
| `cp -v` | *verbose* (detallado) | Muestra en pantalla lo que va copiando | `cp -v notas.txt copia.txt` |
| `echo` | *echo* (eco) | Muestra un texto en pantalla | `echo "Hola"` |
| `echo >` | `>` = redirigir y sobrescribir | Guarda el texto en un archivo. Si el archivo existe, reemplaza todo su contenido; si no existe, lo crea | `echo "Hola" > notas.txt` |
| `echo >>` | `>>` = redirigir y añadir | Añade el texto al final del archivo, en una línea nueva, sin borrar lo que ya tenía | `echo "Otra línea" >> notas.txt` |
| `echo >>` (sin texto) | Línea vacía | Añade una línea en blanco al final del archivo | `echo >> notas.txt` |
| `echo -n` | *no newline* (sin salto de línea) | No agrega el salto de línea al final, así lo siguiente queda en la misma línea | `echo -n "Hola " >> notas.txt` |
| `echo -e` | *enable escapes* (activar caracteres de escape) | Interpreta caracteres de escape como `\n` o `\t` (ver tabla abajo) | `echo -e "Línea 1\nLínea 2" >> notas.txt` |
| `find` | *find* (encontrar) | Busca archivos y carpetas dentro de una ruta y todas sus subcarpetas | `find .` |
| `find -name` | *name* (nombre) | Busca por nombre. Acepta comodines, pero ponlos entre comillas | `find . -name "*.csv"` |
| `find -iname` | *insensitive name* (nombre sin distinguir mayúsculas) | Igual que `-name`, pero no distingue mayúsculas de minúsculas | `find . -iname "*.csv"` |
| `find -type` | *type* (tipo) | Filtra por tipo: `f` para archivos y `d` para carpetas | `find . -type d` |
| `grep` | *global regular expression print* (buscar e imprimir coincidencias) | Busca un texto dentro de archivos y muestra las líneas que lo contienen | `grep "Ana" datos.csv` |
| `grep -i` | *ignore case* (ignorar mayúsculas) | Busca sin distinguir mayúsculas de minúsculas | `grep -i "ana" datos.csv` |
| `grep -n` | *line number* (número de línea) | Muestra el número de línea de cada coincidencia | `grep -n "Luis" datos.csv` |
| `grep -v` | *invert* (invertir) | Muestra las líneas que **no** contienen el texto | `grep -v "error" datos.csv` |
| `grep -c` | *count* (contar) | Muestra cuántas líneas coinciden, en lugar de mostrarlas | `grep -c "error" app.log` |
| `grep -r` | *recursive* (recursivo) | Busca en todos los archivos de una carpeta y sus subcarpetas | `grep -r "Ana" .` |
| `head` | *head* (cabeza) | Muestra las primeras 10 líneas de un archivo | `head datos.csv` |
| `head -n` | *number* (número de líneas) | Muestra las primeras *n* líneas | `head -n 5 datos.csv` |
| `head >` | Guardar el inicio | Guarda las primeras líneas en un archivo nuevo. Útil para crear una muestra pequeña de un dataset grande | `head -n 100 datos.csv > muestra.csv` |
| `ls` | *list* (listar) | Muestra el contenido de la carpeta | `ls Documents` |
| `ls -l` | *long* (formato largo) | Muestra detalles: permisos, tamaño, fecha y dueño | `ls -l` |
| `ls -a` | *all* (todo) | Incluye los archivos ocultos (los que empiezan con `.`) | `ls -a` |
| `ls -h` | *human-readable* (legible para humanos) | Junto con `-l`, muestra tamaños en KB, MB o GB | `ls -lh` |
| `ls -la` | *long* + *all* | Combina las dos opciones: detalles e incluye ocultos | `ls -la` |
| `mkdir` | *make directory* (crear directorio) | Crea una carpeta nueva | `mkdir proyecto` |
| `mkdir -p` | *parents* (padres) | Crea también las carpetas intermedias que no existan, y no marca error si la carpeta ya existe | `mkdir -p datos/crudos/2026` |
| `mv` | *move* (mover) | Mueve o renombra archivos y carpetas | `mv notas.txt apuntes.txt` |
| `mv -i` | *interactive* (interactivo) | Pregunta antes de sobrescribir un archivo que ya existe | `mv -i notas.txt datos/` |
| `mv -n` | *no-clobber* (no sobrescribir) | Nunca sobrescribe un archivo existente | `mv -n notas.txt datos/` |
| `mv -v` | *verbose* (detallado) | Muestra en pantalla lo que va moviendo | `mv -v notas.txt datos/` |
| `pwd` | *print working directory* (imprimir el directorio de trabajo) | Muestra la ruta en la que se encuentra | `pwd` → `/c/Users/Rieux` |
| `rm` | *remove* (eliminar) | Elimina archivos. No los manda a la papelera: se borran para siempre | `rm notas.txt` |
| `rm -r` | *recursive* (recursivo) | Elimina una carpeta con todo su contenido | `rm -r carpeta` |
| `rm -i` | *interactive* (interactivo) | Pide confirmación antes de borrar cada archivo | `rm -i notas.txt` |
| `rm -f` | *force* (forzar) | Borra sin pedir confirmación y sin avisar si el archivo no existe | `rm -f notas.txt` |
| `rm -d` | *directory* (directorio) | Elimina una carpeta, solo si está vacía | `rm -d carpeta_vacia` |
| `rm -v` | *verbose* (detallado) | Muestra en pantalla cada archivo que va eliminando | `rm -v notas.txt` |
| `tail` | *tail* (cola) | Muestra las últimas 10 líneas de un archivo | `tail datos.csv` |
| `tail -n` | *number* (número de líneas) | Muestra las últimas *n* líneas | `tail -n 5 datos.csv` |
| `tail -n +` | Desde la línea *n* | Muestra desde la línea *n* hasta el final. Con `+2` se salta el encabezado de un CSV | `tail -n +2 datos.csv` |
| `tail -f` | *follow* (seguir) | Sigue mostrando las líneas nuevas mientras el archivo crece. Útil para ver logs en vivo. Sal con `Ctrl + C` | `tail -f app.log` |
| `tail >` | Guardar el final | Guarda las últimas líneas en un archivo nuevo | `tail -n 100 datos.csv > ultimos.csv` |
| `touch` | *touch* (tocar) | Crea un archivo vacío. Si ya existe, solo actualiza su fecha de modificación | `touch notas.txt` |
| `unalias` | *unalias* (quitar alias) | Elimina un alias | `unalias ll` |
| `wc` | *word count* (contar palabras) | Cuenta líneas, palabras y bytes de un archivo, en ese orden | `wc datos.csv` → `4 5 36 datos.csv` |
| `wc -l` | *lines* (líneas) | Cuenta solo las líneas. Útil para saber cuántas filas tiene un CSV | `wc -l datos.csv` |
| `wc -w` | *words* (palabras) | Cuenta solo las palabras | `wc -w notas.txt` |
| `wc -c` | *bytes* | Cuenta solo los bytes (el tamaño del archivo) | `wc -c datos.csv` |
| `env` |  |  |  |
| `history` |  |  |  |



> **Cuidado:** `rm -rf` combina *recursive* y *force*: borra carpetas completas sin preguntar y sin posibilidad de recuperarlas. Revisa bien la ruta antes de usarlo.

### Caracteres de escape

Son combinaciones que empiezan con `\` (barra invertida) y representan caracteres especiales. Con `echo` hay que usar la opción `-e` para que se interpreten.

| Carácter | Significado | Qué hace | Ejemplo |
| --- | --- | --- | --- |
| `\n` | *new line* (nueva línea) | Salto de línea | `echo -e "Hola\nMundo"` |
| `\t` | *tab* (tabulación) | Inserta una tabulación | `echo -e "Nombre\tEdad"` |
| `\\` | Barra invertida | Escribe una `\` literal. Evita que `\n` se lea como salto de línea | `echo -e 'C:\\nuevo'` → `C:\nuevo` |
| `\"` | Comillas dobles | Escribe comillas dentro de un texto entre comillas dobles (funciona sin `-e`) | `echo "Dijo \"hola\""` |

> **Comillas simples y dobles:** entre comillas dobles, bash reemplaza variables como `$HOME` por su valor. Entre comillas simples, todo se escribe tal cual: `echo '$HOME'` muestra el texto `$HOME`.

### Rutas absolutas y relativas

- **Ruta absoluta**: empieza desde la raíz (`/`) o desde tu *home* (`~`). Funciona sin importar dónde estés. Ejemplo: `cd ~/Documents/GitHub`
- **Ruta relativa**: parte de la carpeta en la que estás (la que muestra `pwd`). Ejemplo: `cd GitHub`

> Tip: si una carpeta tiene espacios en el nombre, ponla entre comillas: `cd "Data Scientist"`

## Comodines (Wildcards)

Los comodines son caracteres especiales que sirven para seleccionar varios archivos a la vez según un patrón, en lugar de escribir cada nombre. Funcionan con casi cualquier comando: `ls`, `cp`, `mv`, `rm`, `cat`, etc.

Para los ejemplos, imagina una carpeta con: `dato1.csv`, `dato2.csv`, `datoA.csv`, `dato10.csv` y `notas.txt`.

| Comodín | Nombre | Qué hace | Ejemplo |
| --- | --- | --- | --- |
| `*` | *asterisk* (asterisco) | Representa cualquier cantidad de caracteres, incluso ninguno | `ls *.csv` → los cuatro archivos `.csv` |
| `?` | *question mark* (signo de interrogación) | Representa exactamente un carácter | `ls dato?.csv` → `dato1.csv`, `dato2.csv`, `datoA.csv` (no incluye `dato10.csv`) |
| `[ ]` | *brackets* (corchetes) | Representa un solo carácter de los que están dentro | `ls dato[12].csv` → `dato1.csv`, `dato2.csv` |
| `[a-z]` / `[0-9]` | Rango dentro de corchetes | Representa un carácter dentro del rango | `ls dato[0-9].csv` → `dato1.csv`, `dato2.csv` |
| `[! ]` | Corchetes con negación | Representa un carácter que **no** esté dentro | `ls dato[!0-9].csv` → `datoA.csv` |
| `{ }` | *curly braces* (llaves) | Genera varias palabras a partir de una lista separada por comas | `mkdir {datos,notebooks,reportes}` → crea las tres carpetas |
| `{1..5}` | Rango dentro de llaves | Genera una secuencia de números o letras | `touch dia{1..5}.txt` → `dia1.txt` a `dia5.txt` |

> **Diferencia clave:** `*`, `?` y `[ ]` buscan archivos que **ya existen**. Las llaves `{ }` no buscan nada: solo generan texto, por eso sirven para **crear** archivos o carpetas nuevos.

> **Se pueden combinar:** `ls *.{csv,txt}` muestra todos los archivos `.csv` y `.txt`.

> **Cuidado:** los comodines no incluyen archivos ocultos (los que empiezan con `.`). Y úsalos con cuidado junto a `rm`: `rm *` borra todos los archivos de la carpeta. Antes de borrar, prueba el patrón con `ls` para ver qué archivos selecciona.

## Entornos virtuales con `uv`

Un **entorno virtual** es una carpeta con su propia versión de Python y sus propias librerías, separada del resto de tu computadora. Así cada proyecto tiene las librerías que necesita, sin afectar a los demás.

Los comandos se escriben en la terminal (Git Bash o PowerShell), no en el notebook. En los ejemplos, el entorno se llama `mi_entorno`.

| Tarea | Comando | Qué hace |
| --- | --- | --- |
| Crear un entorno | `uv venv mi_entorno` | Crea el entorno en una carpeta llamada `mi_entorno`. Si solo escribes `uv venv`, lo crea en una carpeta llamada `.venv`, que es el nombre más común |
| Crear un entorno con una versión de Python | `uv venv mi_entorno --python 3.12` | Crea el entorno con la versión de Python que indiques |
| Crear un entorno a partir de un archivo de requerimientos | `uv venv mi_entorno` y después `uv pip install -r requirements.txt --python mi_entorno` | Primero crea el entorno y luego instala todo lo que está en el archivo. `-r` significa *requirements* (requerimientos) |
| Activar un entorno (Git Bash) | `source mi_entorno/Scripts/activate` | A partir de aquí, `python` y `uv pip` usan ese entorno. El nombre del entorno aparece en el prompt |
| Activar un entorno (PowerShell) | `mi_entorno\Scripts\Activate.ps1` | Lo mismo, pero en PowerShell |
| Desactivar un entorno | `deactivate` | Regresa al Python normal de tu computadora. Funciona igual en Git Bash y PowerShell |
| Instalar librerías en el entorno activo | `uv pip install pandas seaborn` | Instala una o varias librerías en el entorno que esté activado |
| Instalar librerías en un entorno específico | `uv pip install pandas --python mi_entorno` | Instala en el entorno que indiques, sin necesidad de activarlo |
| Ver las librerías instaladas | `uv pip list` | Muestra las librerías del entorno activo. Agrega `--python mi_entorno` para ver las de otro entorno |
| Guardar las librerías en un archivo | `uv pip freeze > requirements.txt` | Crea el archivo de requerimientos con las librerías y versiones del entorno, para que otra persona pueda recrearlo |
| Eliminar un entorno | `rm -rf mi_entorno` | Borra la carpeta del entorno. `uv` no tiene un comando para esto porque el entorno es solo una carpeta. Desactívalo antes de borrarlo |

> **Diferencia con conda:** conda guarda todos los entornos en una carpeta central y se eligen por nombre. Con `uv`, el entorno es una carpeta dentro de tu proyecto, así que se indica por su ruta.

> **Tip:** agrega la carpeta del entorno a tu `.gitignore` para no subirla a GitHub. Lo que se comparte es el `requirements.txt`.
