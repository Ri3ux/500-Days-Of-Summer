# Teoría de la probabilidad
Para empezar, vamos a definir algunos términos clave. Imagina un escenario en el que tiras un dado y tratas de obtener un resultado de 5 o superior. Aquí tienes algunos términos para describir esta situación:

| Término | Definición | Ejemplo |
|---|---|---|
| Experimento | Cualquier procedimiento que puede ser repetido. | Tirar un dado. |
| Prueba | Una repetición de un experimento. | Cada intento de tirar un dado. |
| Resultado | Resultado de un experimento que se ejecuta una vez. | Tirar un 1, 2, 3, 4, 5 o 6. |
| Evento | Resultado único o conjunto de resultados a los que se puede asignar una probabilidad. | Tirar 5 o 6 (cualquier resultado posible que sea igual o mayor que 5). |

## La fórmula principal
Cuando tiras un dado, cada posible resultado tiene la misma probabilidad de ocurrir. Como hay seis resultados posibles y cada uno es igualmente probable, la probabilidad de cada resultado es 1/6.

En estadística, utilizamos la letra P para denotar probabilidad y la letra A para denotar un evento de nuestro interés. Por lo tanto, la notación P(A) describe la probabilidad del evento A.

El P(A) puede calcularse del siguiente modo:

$
P(A) = \frac{\#\text{ de resultados que satisfacen el evento}}{\#\text{ de resultados totales}}
$

Si el evento A se define como "tirar un número de 5 o mayor", entonces:

$
P(A) = \frac{1+1}{6} = \frac{1}{3}
$

Muchos otros escenarios tienen esta propiedad. Por ejemplo, al sacar una sola carta al azar de una baraja de 52 cartas, es igualmente probable que se saque cada carta. Si el evento A se define como "obtener el ocho de corazones" (o cualquier otra carta específica), entonces la probabilidad P(A) es 1/52.

Podemos generalizar este escenario. Si…

un experimento se ejecuta una vez,
la probabilidad de cada resultado es la misma, y
hay n resultados posibles,
entonces la probabilidad de que ocurra cada resultado es 1/n.

Esto nos lleva a una propiedad esencial de la probabilidad: la suma de las probabilidades de todos los resultados posibles en un experimento debe ser igual al 100% o a 1. Esto asegura que se producirá uno de los resultados posibles.

Al lanzar un dado, hay seis resultados posibles, cada uno con una probabilidad de ⅙. La suma de estas probabilidades es ⅙ + ⅙ + ⅙ + ⅙ + ⅙ + ⅙ = 1. Por lo tanto, se garantiza que sacaremos un 1, 2, 3, 4, 5, o 6.

## Eventos y espacios muestrales
El conjunto de todos los resultados posibles de un experimento se denomina espacio muestral, normalmente denotado con `S`.

Para ilustrarlo, consideremos el ejemplo de lanzar un solo dado. El espacio de muestra asociado a este evento es:

$
S = \{1, 2, 3, 4, 5, 6\}
$

Exploremos cómo calcular la probabilidad de un evento a partir del espacio muestral.

Por ejemplo, supongamos que el evento A es "sacar un número menor que 3". Después de examinar el espacio muestral, observamos que hay dos números que son menores que 3. Por consiguiente, la probabilidad del evento A, representada por P(A), donde P representa la probabilidad, es 2 (el número de resultados que satisfacen A) dividido entre 6 (la longitud de S).

$
P(A) = \frac{2\text{ (número de resultados que satisfacen A)}}{6\text{ (longitud de S)}} = \frac{2}{6}
$

## La ley de los grandes números
Cada evento está asociado a una probabilidad. En sistemas complejos o poblaciones grandes, calcular la probabilidad real de un evento puede ser muy difícil o incluso imposible. Aquí es cuando confiamos en las estimaciones.

Al estimar la probabilidad de un evento, nos basamos en un teorema muy útil: la **ley de los grandes números**. Esta ley establece que cuantas más veces se repita un experimento, más cerca estará la frecuencia relativa de un evento determinado (el número de veces que ocurre el evento dividido entre el número de veces que se repite el experimento) de su verdadera probabilidad.

Por ejemplo, ya sabemos que al lanzar una moneda al aire, la probabilidad de obtener cara es del 50%. Sin embargo, si no tuviéramos esta información, podríamos estimar la probabilidad lanzando una moneda al aire muchas veces y rastreando el número de caras y el número de lanzamientos.

La ley de los grandes números establece que, a medida que el número de lanzamientos se acerca al infinito, la frecuencia relativa de este evento se aproximará a la verdadera probabilidad del evento.

Las probabilidades de eventos complejos se estiman realizando experimentos muchas veces y utilizando frecuencias relativas para estimar las probabilidades.

## Intersección de eventos
Es posible que un resultado se asocie con múltiples eventos. Por ejemplo, los siguientes eventos comparten resultados:

* *Evento A* = la suma de dos dados sea mayor a `8`.
* *Evento B* = cada dado mostrará el mismo número.



Ten en cuenta que si cada dado muestra `5` o `6`, entonces ocurren tanto `A` como `B`.

### Diagramas de Venn
Para ilustrar la ocurrencia de ambos eventos (como `A` y `B` arriba cuando cada dado mostró `5` o `6`), podemos usar un **diagrama de Venn**. No muestra resultados elementales, solo la intersección relevante entre los eventos:



Los eventos que no pueden ocurrir simultáneamente en el mismo experimento se denominan **mutuamente excluyentes**; no se cruzan en el diagrama de Venn:



La probabilidad de la intersección de eventos mutuamente excluyentes es 0.

Si los eventos `A` y `B` son mutuamente excluyentes, significa que P(A and B) = 0.

$
P(A \text{ and } B) = 0
$

Como puedes ver arriba, si sucede el evento A, no hay forma de que el evento B suceda también.

Si los eventos mutuamente excluyentes llenan todo el espacio muestral, entonces la suma de sus probabilidades será igual a `1`.


Por ejemplo, consideremos los siguientes eventos:

* Dos dados muestran un número igual (celdas rojas en la imagen de abajo).
* El primer dado mostró más que el segundo (celdas verdes).
* El segundo dado mostró más que el primero (celdas azules).



En estos eventos se incluyen todos los resultados elementales; cubren todo el espacio muestral. Sus probabilidades son `6/36`, `15/36` y `15/36`, respectivamente. Juntos suman 1: (6 + 15 + 15) / 36 = 1.

### Eventos independientes y dependientes
Los eventos se denominan **independientes** si la ocurrencia de uno no afecta la probabilidad del otro.

Si el evento A = *John paga $10 extra por una garantía de 1 año al finalizar la compra* y B = *Sophie hace clic en ese enlace de cebo de clics*, entonces si John paga por la garantía, no afectará la elección de Sofía. Podemos suponer que estos dos eventos son independientes.

Si el evento A = *llueve* y B = *usas un chubasquero para ir a trabajar*, entonces la probabilidad de B probablemente dependerá de la probabilidad de A. Estos eventos son eventos **dependientes**.

Con un diagrama de Venn, es fácil ver si los eventos son mutuamente excluyentes. Sin embargo, no es tan fácil determinar su independencia; necesitas verificar si el producto de las probabilidades de los eventos es igual a la probabilidad de su intersección. Para encontrar la probabilidad de intersección entre eventos independientes (de que ocurran ambos eventos), necesitamos encontrar el producto de las probabilidades de cada evento.

#### Probabilidades de eventos independientes
Esto es lo bueno de los eventos independientes…

Si los eventos *A* y *B* son independientes, entonces:

$
P(A \text{ and } B) = P(A)×P(B)
$

Esto significa que solo necesitas conocer sus probabilidades individuales para calcular la probabilidad de que ocurran ambas.

<div class="alert alert-block alert-danger">
<b>Recuerda que...</b> <a class="tocSkip"></a>

Esto NO es cierto para los eventos dependientes.
</div>

##### Ejemplo de eventos independientes
Que los eventos A y B sean igual a…

* A = El primer dado muestra 3.
* B = El segundo dado muestra 5.
  
Estos eventos son independientes; el resultado del primer dado no tiene ningún efecto sobre el segundo. Dado que es igualmente probable que cada tirada de dado dé como resultado un valor de 1 a 6, sabemos que P(A) = ⅙ y P(B) = ⅙.

Siguiendo la fórmula, podemos calcular:

$
P(A \text{ and } B) = \frac{1}{6}×\frac{1}{6}=\frac{1}{36}
$

Visualmente, vemos que este cálculo es verdadero al observar solo 1 resultado de los 36 donde A y B están satisfechos.



##### Ejemplo de eventos dependientes
¿Los eventos "uno de los dados muestra 2" y "los dados muestran 7 en total" son independientes en el espacio muestral?

$
P(A)\text{ (probabilidad de que los dados muestren 7 en total) }=\frac{6}{36}=\frac{1}{6}
$

$
P(B)\text{ (probabilidad de que uno de los dados muestre 2) }=\frac{11}{36}\text { (como 2,2 aparece en ambas listas, no podemos contarlo dos veces, por eso no es 12/36) }
$

$
P(A∩B)\text{ (probabilidad de intersección) }=\frac{2}{36}
$

Tenemos que:
$
P(A∩B)=P(A)P(B)
$

Entonces:
$
\frac{2}{36}=\frac{1}{6}×\frac{11}{36}
$

La igualdad no se cumple:
$
\frac{12}{216}≠\frac{11}{216}
$

La ocurrencia de uno de estos eventos cambia la probabilidad del otro evento.

## Fundamentos de combinatoria y otros problemas de probabilidad
### Permutaciones
Para calcular el número de permutaciones de n elementos, los matemáticos utilizan la fórmula:

$
P_n = n!
$

Es posible que no hayas encontrado antes el signo `!` en expresiones matemáticas. Este signo se llama **factorial** y significa el producto de números naturales de `1` a `n`. En otras palabras,

$
n! = 1 \cdot 2 \cdot 3 \cdot \ldots \cdot (n - 1) \cdot n
$

Así, para tres cursos diferentes, hay 3! = 1 × 2 × 3 = 6 itinerarios educativos diferentes.

#### Calcular factoriales en Python
Cuando el valor de `n` es pequeño, `n!` puede calcularse mentalmente. Para valores mayores de `n`, resulta más cómodo automatizar el proceso. Para ello, puedes utilizar la función `factorial()` del módulo `math` de la librería estándar de Python.

````python
# Importa la función factorial del módulo matemático
from math import factorial

# Define el número necesario de cursos
courses_amount = 3

# Calcula el factorial de 3 (el valor de la variable courses_amount)
result = factorial(courses_amount)

# Muestra el resultado
print('El factorial de 3 es:', result)
````

### Combinaciones
Esta es una nueva tarea para ti, y antes de resolverla, tenemos que aprender una nueva fórmula: la fórmula para calcular el número de combinaciones de `k` elementos de `n` opciones posibles. La fórmula es la siguiente:

$
C_n^k = \frac{n!}{k!(n-k)!}
$

Por ejemplo, si una cafetería vende 10 tipos de helado y quieres comprar tres diferentes, tienes:

$
C_{10}^{3} = \frac{10!}{3!(10-3)!} = \frac{10!}{3!7!} = \frac{1\cdot2\cdot3\cdot4\cdot5\cdot6\cdot7\cdot8\cdot9\cdot10}{(1\cdot2\cdot3)(1\cdot2\cdot3\cdot4\cdot5\cdot6\cdot7)} = 120
$

Lo mismo puede calcularse en Python:

````python
from math import factorial
n = 10
k = 3

combinations = factorial(n)/(factorial(k) * factorial(n-k))
print('La cantidad de helados con 3 sabores diferentes es:', combinations)
````

## La idea de distribución
Ahora que sabes cómo se relacionan los experimentos, los resultados y los eventos con las probabilidades, estamos listos para presentar el concepto de una **variable aleatoria**.

### Variables aleatorias
Las variables aleatorias son aquellas que toman valores aleatorios. El término "aleatorio" en este caso significa que no podemos predecir con un 100% de exactitud qué valor tomará la variable.

Por ejemplo, digamos que `X` es la altura de la próxima persona que entre por tu puerta. `X` es una variable aleatoria con un rango de valores posibles. El valor real que tomará `X` quedará desconocido hasta que alguien entre por la puerta. En ese momento, se observará el valor de `X`.

En la teoría de la probabilidad, nos interesa conocer todos los valores posibles que puede tomar `X` y las probabilidades asociadas a estos valores.

````python
import numpy as np

exam_results = np.array(
    [
        42,  56,  59,  76,  43,  34,  62,  51,  50,  65,  
        66,  50,  46,  5,  79, 99,  51,  26,  35,   8,  
        34,  47,  64,  58,  61,  12,  30,  63,  20,  68
    ]
)

summarized_data = {'excellent': 0,
                   'good': 0,
                   'average': 0,
                   'passable': 0,
                   'failed': 0}

# escribe aquí el código para contar el número de estudiantes para cada tipo de resultados obtenidos.

# código para mostrar los resultados en pantalla. No lo cambies.
for grade in exam_results:
    if grade >= 90:
        summarized_data['excellent'] += 1
    elif grade >= 70:
        summarized_data['good'] += 1
    elif grade >= 50:
        summarized_data['average'] += 1
    elif grade >= 20:
        summarized_data['passable'] += 1
    else:
        summarized_data['failed'] += 1

for result in summarized_data:
    print(result, '-', summarized_data[result])
````

## Distribución binomial
En el mundo real, hay muchos experimentos que tienen dos resultados posibles. Por ejemplo, ganar o no ganar la lotería; o si una tostada con mantequilla cae al suelo por la parte untada o no; o si un cliente cambia su cuenta a una cuenta premium o rechaza la oferta. Cuando realizamos un experimento una vez y hay dos resultados posibles, estamos ante un **experimento binomial simple** o **ensayo de Bernoulli**.

Normalmente, llamamos a estos dos resultados *éxito* y *fracaso*. Si la probabilidad de éxito es `p`, entonces la probabilidad de fracaso debe ser igual a `(1-p)`, ya que solo hay dos resultados posibles y deben sumar `1`.

La distribución de probabilidad descrita aquí se conoce como distribución binomial.

### Experimentos binomiales
Si realizamos un experimento con dos resultados posibles un número fijo de veces, obtendremos un **experimento binomial**. Por ejemplo, si lanzamos una moneda 100 veces, puede que queramos saber qué probabilidad hay de obtener exactamente 50 caras y 50 cruces.

Si las pruebas se realizan de forma independiente, entonces la probabilidad de una cierta combinación de éxitos y fracasos es el producto de las probabilidades de cada éxito y fracaso individual.

Por ejemplo, para determinar la probabilidad de la combinación *éxito-éxito-fracaso-éxito* **en ese orden** (en lugar de simplemente "tres éxitos y un fracaso"), calculamos el producto $p \times p \times (1-p) \times p$, donde $p$ es la probabilidad de éxito.

La tarea comienza a complicarse si no nos preocupamos por el orden exacto de los éxitos y fracasos.

### Calcular el número de *k* éxitos de *n* pruebas

Como recordarás, tenemos que encontrar la manera de calcular de cuántas formas podemos obtener *k* éxitos de *n* pruebas sin necesidad de compilar una tabla con todos los resultados posibles. Es posible con la ayuda de la siguiente fórmula:

$C_n^k = \frac{n!}{k!(n-k)!}$

Si aplicamos esta fórmula al ejemplo de los clics en el banner, podemos considerar *n* como el número total de usuarios y *k* como el número de clics en el banner. La fórmula nos ayudará a calcular de cuántas formas podemos seleccionar un grupo de *k* usuarios del total de *n* usuarios que hagan clic en el banner y no en otro enlace.

Por ejemplo, el número de formas en que 7 de cada 10 usuarios harán clic en el banner es:

$C_{10}^{7} = \frac{10!}{7!(10-7)!} = \frac{10!}{7!3!} = 120$

Para encontrar el mismo resultado en Python podemos utilizar el código:

````python
from math import factorial
n = 10
k = 7

combinations = factorial(n)/(factorial(k) * factorial(n-k))

print('El número de formas en que 7 de cada 10 usuarios hagan clic sobre el banner es de', combinations)
````

### Encontrar la probabilidad de *n* éxitos

Bien, podemos calcular cuántas formas hay de obtener *k* éxitos a partir de *n* pruebas. Pero, ¿cómo podemos calcular la probabilidad de obtener ese número de éxitos?

Recuerda que, aunque habrá muchas combinaciones que den *k* éxitos, cada una de ellas tendrá la misma probabilidad. Por lo tanto, necesitamos:

1. Encontrar la probabilidad de una sola forma de obtener *k* éxitos de *n* experimentos.
2. Encontrar el número total de combinaciones que dan *k* éxitos.
3. Multiplicar ambos resultados.

Veamos más detenidamente el primer punto. Supongamos que los 7 primeros clics se producen en el banner y los 3 restantes no. La probabilidad de este resultado es:

$0.88 \cdot 0.88 \cdot 0.88 \cdot 0.88 \cdot 0.88 \cdot 0.88 \cdot 0.88 \cdot 0.12 \cdot 0.12 \cdot 0.12 = 0.88^7 \cdot 0.12^3 \approx 0.0007$

En el caso general, la fórmula es la siguiente:

$P = p^k \cdot q^{n-k}$

donde:

- *p* es la probabilidad de éxito.
- *q* es la probabilidad de fracaso.
- *k* es el número de pruebas realizadas con éxito.
- *n* es el número total de pruebas.

Por lo tanto, podemos ver que la probabilidad de obtener *k* éxitos de *n* pruebas es:

$P = C_n^k \cdot p^k \cdot q^{n-k}$

Ahora sabes cómo calcular la probabilidad de obtener un número fijo de éxitos cuando conocemos el número de pruebas y la probabilidad de éxito, así cómo crear una distribución de probabilidad completa para cada número posible de éxitos: de 0 a *n*. Esta distribución se llama **distribución binomial**.

## Distribución normal
La **distribución normal**, a menudo referida como la **campana de Gauss** se manifiesta en diversas situaciones, desde representar la altura de la población hasta determinar las tasas de falla en equipos.

Aunque existen otras distribuciones en el mundo real, la normal se destaca por su prevalencia y relevancia teórica.

### Centro y dispersión de una distribución normal

Una distribución normal tiene dos parámetros clave: la media y la varianza.

$X \sim (\mu, \sigma^2)$

Esta notación se puede leer como: la variable *X* se distribuye normalmente con una media mu ($\mu$) y una varianza sigma al cuadrado ($\sigma^2$), es decir, una desviación estándar sigma.

Para una distribución normal, el 68% de los puntos de datos se encuentran a una desviación estándar ($\sigma$) de la media, el 95% a dos desviaciones estándar de la media y el 99,7% a las tres primeras desviaciones estándar de la media. Este concepto se denomina **regla de las 3 sigmas**.



#### Calcular probabilidades en datos normalmente distribuidos con Python

SciPy es una librería Python gratuita y de código abierto que se utiliza para la computación científica y técnica. Contiene módulos para optimización, álgebra lineal, integración, procesamiento de señales e imágenes y muchos otros propósitos.

En esta lección examinaremos la función `stats.norm()`, que genera una distribución normal con unos parámetros dados (media y desviación estándar). También estudiaremos los métodos `norm.cdf` y `norm.ppf`, que permiten buscar respuestas a preguntas concretas.

Veamos estos métodos con más detalle. Para ilustrarlo, consideremos el siguiente ejemplo: en la India, los estudiantes mayores de edad gastan una media de 5000 dólares para aprender una nueva profesión, con una desviación típica de 1500 dólares.

| Nombre | Sintaxis | Explicación | Ejemplo |
|---|---|---|---|

| La función de punto porcentual | `norm.ppf()` | Da el valor de la variable aleatoria que corresponde a una determinada probabilidad | ¿Cuál es el coste máximo de la formación para el 10% de los estudiantes que gastaron menos dinero en sus estudios? |

````python
# Instalamos la librería dentro del documento
%pip install scipy
````

````python
# Importamos el módulo stats.
from scipy import stats as st

# Creamos un objeto con datos distribuidos normalmente con una media de 5000
# Y una desviación estándar de 1500.

data = st.norm(5000, 1500)

# Creamos una variable para almacenar el coste deseado.
desired_cost = 4000

# Calculamos la probabilidad de obtener el valor desired_cost.
# cdf significa Cumulative Distribution Function, o función de distribución acumulada.

probability = data.cdf(desired_cost)
probability = round(probability * 100, 2)


print(f'Hay aproximadamente un {probability}% de probabilidad de que el costo sea de $4.000 o menos.')
````

## Combinación de las distribuciones binomial y normal
### Valor esperado
Digamos que tenemos una variable aleatoria. Puede tomar ciertos valores y cada valor tiene una probabilidad asociada. El **valor esperado** de la variable aleatoria es el valor "típico" que asociamos con esta variable aleatoria. Para encontrar este valor, necesitamos repetir el experimento muchas veces para aproximar las probabilidades de cada valor y, luego, encontrar el promedio ponderado.

Un experimento binomial consta de n pruebas de un ensayo simple (de Bernoulli) con dos resultados. Si el número de pruebas es lo suficientemente alto, la distribución binomial se puede modelar mediante la distribución normal. En esta lección, te mostraremos cómo funciona.

#### Ejemplos de valor esperado

Te presentamos a Archie. Archie es un ingeniero de software que escribe 50 fragmentos de código cada día. Una de cada cinco veces, el fragmento se ejecuta perfectamente a la primera.

Esto es un experimento binomial diario con:

$n = 50 \quad p = 0.2$

**Valor esperado**

$E[X] = n \cdot p = 50 \cdot 0.2 = 10$

**Varianza**

$Var(X) = n \cdot p \cdot (1-p) = 50 \cdot 0.2 \cdot 0.8 = 8$

**Desviación estándar**

$\sigma = \sqrt{8} = 2\sqrt{2} \approx 2.83$

---

Esto no significa que Archie escriba 10 fragmentos perfectos todos los días. Un día podría estar cansado y escribir solo 5; otro día, sentirse genial y escribir 15.

Pero con el tiempo, el promedio diario se aproximará a 10, y aproximadamente el 99% de los valores estarán dentro de:

$10 \pm 3\sigma = 10 \pm 3(2\sqrt{2}) \approx 10 \pm 8.49$

Es decir, en un día cualquiera se puede esperar que **entre 2 y 18** fragmentos se ejecuten perfectamente en el primer intento.


### Aproximación normal a la distribución binomial
Si hay 50 intentos o más, los resultados de un experimento binomial pueden mostrarse como datos con una distribución normal. Combinar dos tipos de distribuciones de probabilidad tan diferentes puede parecer confuso, así que veamos esta idea con más detalle.

Consideremos una distribución binomial con $n = 50$ y $p = 0.8$. Al aplicar la fórmula para la probabilidad de obtener $k$ éxitos de $n$ pruebas, obtendremos los siguientes resultados:

* La probabilidad de obtener 30 éxitos de 50 pruebas es de 0.0006.
* La probabilidad de obtener 35 éxitos de 50 pruebas es de 0.0299.
* La probabilidad de obtener 40 éxitos de 50 pruebas es de 0.1398.
* La probabilidad de obtener 45 éxitos de 50 pruebas es de 0.0295.

Si calculamos los resultados para cada número de éxitos y creamos un gráfico basado en ellos, obtendremos un gráfico muy parecido a la curva de distribución normal.



¿Por qué es útil? Porque la distribución normal tiene características útiles y predecibles, como la regla de las tres sigmas.

#### Ejercicio 1
Una empresa ha decidido anunciarse online. El servicio de publicidad dice que, de promedio, el 15% de los usuarios hacen clic en sus anuncios. Eso es 750 visitas por cada 5000 vistas.

La empresa coloca anuncios, compra 5000 impresiones (vistas) y solo obtiene 715 visitas. ¡El equipo de marketing está disgustado! ¿No les habían prometido 750 visitas? Usaremos argumentos estadísticos para calmarlos.

En resumen: cuando un usuario ve el anuncio, hay un 15% de posibilidades de que haga clic en él. Esto nos da una distribución binomial con $n = 5000$ y $p = 0.15$.

Utilizaremos la distribución normal para aproximar la distribución binomial y calcular la probabilidad de obtener 715 clics o menos.

Piensa: tenemos nuestro valor del eje X y queremos la probabilidad de obtener ese número o menos. ¿Qué método necesitamos?

¿Ya lo tienes?

Necesitamos `norm.cdf()`:

````python
# Importamos las librerías
from scipy import stats as st
import math as mt

# Establecemos los valores de los parámetros
binom_n = 5000
binom_p = 0.15
clicks = 715

# Calculamos el valor esperado de éxitos y la desviación estándar
mu = binom_n*binom_p
sigma = mt.sqrt(binom_n * binom_p * (1 - binom_p))

# Calcuiamos la probabilidad de obtener 715 vistas o menos
p_clicks = st.norm(mu, sigma).cdf(clicks)
p_clicks = round(p_clicks, 3) * 100

print(f'La probabilidad de obtener {clicks} o menos es de {p_clicks}%')
````

La probabilidad de conseguir 715 visitas en lugar de 750 no es muy alta, alrededor del **8.3%**. Sin embargo, no es lo suficientemente pequeña como para que este evento casi nunca suceda.

Teniendo en cuenta esta probabilidad, podemos afirmar que por cada 100 empresas que pusieron anuncios con el servicio de publicidad, ocho recibirán 715 clics o menos. Resulta que este no es un evento raro en absoluto y que la queja del equipo de marketing no es razonable.

# Guía de Distribuciones de Probabilidad: Binomial y Normal

## 1. ¿Qué es una distribución de probabilidad?

Es una función que describe qué tan probable es cada resultado posible de un experimento aleatorio. Antes de ver las fórmulas, hay que entender un concepto clave: **la frecuencia relativa**.

Si repites un experimento $N$ veces y cuentas cuántas veces obtienes un resultado específico ("éxito"), la frecuencia relativa es:

$\text{frecuencia relativa} = \dfrac{\text{número de éxitos}}{N}$

**Ejemplo:** si eliges una canción al azar de una lista de 100 (donde 10 canciones son "éxito") y repites esto $N$ veces:
- Con $N$ pequeño, el resultado varía mucho por azar.
- Con $N$ grande, la frecuencia relativa se estabiliza cerca de la probabilidad teórica ($10/100 = 0.1$).

Esto se llama **ley de los grandes números**: a más repeticiones, más se acerca lo observado a lo teórico. Es la base intuitiva sobre la que se construyen las distribuciones formales.

---

## 2. Distribución Binomial

### ¿Cuándo se usa?

Cuando cuentas cuántos **éxitos** ocurren en un número fijo de intentos **independientes**, donde cada intento solo tiene dos resultados posibles (éxito/fracaso) y la probabilidad de éxito es siempre la misma.

**Condiciones que debe cumplir el experimento:**
1. Número fijo de intentos ($n$).
2. Cada intento es independiente de los demás.
3. Solo dos resultados posibles por intento (éxito o fracaso).
4. La probabilidad de éxito ($p$) es constante en todos los intentos.

**Ejemplos típicos:**
- Lanzar una moneda 10 veces y contar caras.
- El porcentaje de productos defectuosos en un lote.
- Cuántos folletos promocionales generan una visita a la tienda.

**Cuándo NO aplica:** si las probabilidades cambian entre intentos (por ejemplo, sacar cartas de una baraja *sin devolverlas*), no es binomial. Ahí cada extracción cambia la composición del mazo, así que las probabilidades no son constantes.

### Parámetros

| Símbolo | Significado |
|---|---|
| $n$ | número de intentos |
| $p$ | probabilidad de éxito en cada intento |
| $k$ | número de éxitos que se busca |

### Fórmula principal

$$P(X=k) = \binom{n}{k}p^k(1-p)^{n-k}$$

También se escribe como $P = C_n^k \cdot p^k \cdot q^{n-k}$, donde $q = 1-p$. Son notaciones equivalentes.

**¿Qué hace cada parte?**
- $\binom{n}{k}$: de cuántas formas distintas puedes ordenar $k$ éxitos entre $n$ intentos.
- $p^k$: probabilidad de que ocurran exactamente esos $k$ éxitos.
- $(1-p)^{n-k}$: probabilidad de que los intentos restantes sean fracasos.

**Caso especial:** cuando $k = n$ (todos son éxito), la fórmula se simplifica porque $\binom{n}{n}=1$, quedando solo $p^n$. Es simplemente multiplicar $p$ por sí mismo $n$ veces.

### Media, varianza y desviación estándar

$$E[X] = np \qquad Var(X) = np(1-p) \qquad \sigma = \sqrt{np(1-p)}$$

**Ejemplo resuelto (Archie, el programador):**

Archie escribe $n=50$ fragmentos de código al día, con $p=0.2$ de que cada uno funcione a la primera.

$$E[X] = 50 \times 0.2 = 10$$
$$Var(X) = 50 \times 0.2 \times 0.8 = 8 \qquad \sigma = \sqrt{8} = 2\sqrt{2} \approx 2.83$$

Con la regla empírica, ~99% de los días el número de aciertos estará dentro de $10 \pm 3\sigma \approx 10 \pm 8.49$, es decir entre 1.51 y 18.49. Como los aciertos son números enteros, se redondea al entero alcanzable más cercano dentro del rango: **entre 2 y 18 fragmentos**.

### Código en Python

```python
from scipy.stats import binom

n, p = 10, 0.5

# P(X = k): probabilidad de un valor exacto
binom.pmf(k=3, n=n, p=p)

# P(X <= k): probabilidad acumulada
binom.cdf(k=3, n=n, p=p)

# valor k que deja acumulada una probabilidad q (inversa de cdf)
binom.ppf(q=0.95, n=n, p=p)

# media y varianza
media, var = binom.stats(n, p, moments='mv')

# generar muestras aleatorias
binom.rvs(n=n, p=p, size=1000)
```

## 3. Distribución Normal

### ¿Cuándo se usa?

Para variables **continuas** que tienden a agruparse simétricamente alrededor de un valor central (la media), con menos frecuencia mientras más te alejas de ese centro. Su gráfico es la clásica "campana".

**Ejemplos típicos:**
- Estatura de personas.
- Tiempo de respuesta de un servidor.
- Puntajes de un examen estandarizado.

### Parámetros

| Símbolo | Significado |
|---|---|
| $\mu$ | media (centro de la campana) |
| $\sigma$ | desviación estándar (qué tan dispersos están los datos) |

### Fórmula de densidad

$$f(x) = \dfrac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

A diferencia de la binomial, esta función **no da directamente una probabilidad** al evaluar un punto $x$: da la *densidad* en ese punto. Para obtener probabilidades reales se usa el área bajo la curva, es decir, la función acumulada (`cdf`).

### Código en Python

```python
from scipy.stats import norm

mu, sigma = 0, 1  # normal estándar

# f(x): densidad en un punto (no es una probabilidad directa)
norm.pdf(x=1.5, loc=mu, scale=sigma)

# P(X <= x): probabilidad acumulada
norm.cdf(x=1.5, loc=mu, scale=sigma)

# valor x que deja acumulada una probabilidad q (inversa de cdf)
norm.ppf(q=0.95, loc=mu, scale=sigma)

# generar muestras aleatorias
norm.rvs(loc=mu, scale=sigma, size=1000)
```

---

## 4. `pmf`, `pdf`, `cdf` y `ppf`: cómo no confundirlos

| Función | Para qué distribución | Qué responde |
|---|---|---|
| `pmf(k)` | Discretas (binomial) | "¿Cuál es la probabilidad exacta de obtener el valor $k$?" |
| `pdf(x)` | Continuas (normal) | Densidad en el punto $x$ (no es una probabilidad, es un paso intermedio para calcular áreas) |
| `cdf(x)` | Ambas | "¿Cuál es $P(X \leq x)$?" → das un valor, obtienes una probabilidad acumulada |
| `ppf(q)` | Ambas | La inversa de `cdf`: "¿qué valor de $x$ deja acumulada la probabilidad $q$?" → das una probabilidad, obtienes un valor |

`ppf` es útil para hallar percentiles o valores críticos (por ejemplo, "¿cuál es el valor que deja el 95% de los datos por debajo?").

---

## 5. La aproximación normal a la binomial

### La idea

Cuando $n$ es grande, la forma de una distribución binomial empieza a parecerse mucho a una curva normal. Esto permite usar las fórmulas de la normal (más simples) en vez de la binomial exacta.

**Regla práctica para saber si es válido aproximar:**

$$np \geq 5 \qquad \text{y} \qquad n(1-p) \geq 5$$

Si ambas condiciones se cumplen, la aproximación es confiable.

### ¿Por qué molestarse en aproximar? (el problema real)

Con $n$ pequeño, calcular la binomial exacta es perfectamente manejable. El problema aparece con $n$ grande. Ejemplo: una empresa reparte 5000 folletos, con $p=0.11$ de que cada uno genere una visita, y se quiere saber $P(X<500)$.

Resolverlo con la binomial exacta implicaría:

$$P(X<500) = \sum_{k=0}^{499} \binom{5000}{k}(0.11)^k(0.89)^{5000-k}$$

Esto es problemático por dos razones:
1. **Hay que sumar 500 términos**, cada uno con su propio cálculo de combinatoria.
2. **Los números son extremos.** $\binom{5000}{k}$ involucra factoriales de números enormes (miles de dígitos), combinados con potencias diminutas como $(0.11)^k$. Esto genera errores de precisión numérica incluso en computadoras.

### La solución: usar la normal

Se calculan $\mu$ y $\sigma$ como en la binomial, y luego se usa `norm.cdf` en vez de sumar cientos de términos:

$$\mu = np = 5000 \times 0.11 = 550$$
$$\sigma = \sqrt{np(1-p)} = \sqrt{5000 \times 0.11 \times 0.89} \approx 22.14$$

Se verifica primero que la aproximación es válida:

$$np = 550 \gg 5 \qquad n(1-p) = 4450 \gg 5$$

Y luego, en vez de la suma gigante, un solo cálculo:

```python
from scipy.stats import norm
norm.cdf(x=500, loc=550, scale=22.14)
```

**En resumen:** la binomial exacta es matemáticamente correcta siempre, pero se vuelve computacionalmente costosa e inestable con $n$ grande. La aproximación normal da un resultado prácticamente idéntico con un cálculo mucho más simple y estable.

## 6. Tabla resumen rápida

| Aspecto | Binomial | Normal |
|---|---|---|
| Tipo | Discreta | Continua |
| Parámetros | $n$, $p$ | $\mu$, $\sigma$ |
| Media | $np$ | $\mu$ |
| Varianza | $np(1-p)$ | $\sigma^2$ |
| Desviación estándar | $\sqrt{np(1-p)}$ | $\sigma$ |
| Rango de valores | Enteros: $0, 1, ..., n$ | Cualquier real $(-\infty, \infty)$ |
| Se usa cuando | Cuentas éxitos en $n$ intentos independientes | Mides una variable continua agrupada alrededor de un promedio |

## 7. Consejos rápidos para no confundirte

- Si estás **contando** (número de éxitos en intentos fijos) → binomial.
- Si estás **midiendo** (una magnitud continua) → normal.
- Si las probabilidades cambian entre intentos (por ejemplo, no hay reemplazo) → **no** es binomial, hay que calcular probabilidades sucesivas directamente.
- Con $n$ grande y $np \geq 5$, $n(1-p) \geq 5$ → puedes aproximar la binomial con la normal para simplificar cálculos.
- `pmf`/`pdf` dan la probabilidad (o densidad) de un punto exacto; `cdf` da la probabilidad acumulada hasta un punto; `ppf` hace lo inverso de `cdf`.
